The goal is to now create a function NN, which takes a (flattended?) sample x_in as an input.
Alternatively one could input a collection of samples as x_in

It should then perform y_1 = sigma(W_1*x + b_1) and subsequently y_2 = sigma(W_2*y_1 + b2). Lastly y_out = sigma(W_3*y_2 + b3) should give back a array of length 2, which should give the confidence for the (collection of) sample(s) to be in either of the 2 phases.

##Training

The cost function of this should be sum((y_out - y_correct)^2)

```python
def classifier(x_in, W_arr, b_arr, activation):

    depth_nn = len(b_arr)
    layer_vec_arr = [] #use a list which then contains the np.arrays of the hidden & output layer

    for i in range(depth_nn):
        if i == 0:
            y = activation(W_arr[i] @ x_in + b_arr[i])
            layer_vec_arr.append(y)
        elif i == range(depth_nn)[-1]:
            y_out = softmax(W_arr[i] @ layer_vec_arr[i-1] + b_arr[i])
            layer_vec_arr.append(y_out)
        else:
            y = activation(W_arr[i] @ layer_vec_arr[i-1] + b_arr[i])
            layer_vec_arr.append(y)

    return layer_vec_arr[-1]
```

```python
def init_network_params(sizes):
    rg = np.random.default_rng(1)  # create instance of default random number generator
    W_arr = []
    b_arr = []
    for i in range(len(sizes)-1):
        W_arr.append(rg.uniform(-1,1,(sizes[i+1], sizes[i])))
        b_arr.append(rg.uniform(-1,1,sizes[i+1]))
    return W_arr, b_arr
    
```

```python
W_arr, b_arr = init_network_params((100,25,6,2))
```

```python
size_in = 100
size_h1 = 25
size_h2 = 6
size_out = 2
W1 = rg.random((size_h1, size_in)) - 0.5
W2 = rg.random((size_h2, size_h1)) - 0.5
W3 = rg.random((size_out, size_h2)) - 0.5
W_arr = [W1, W2, W3]

b1 = rg.uniform(-0.5,0.5,size_h1)#rg.random(size_h1) - 0.5
b2 = rg.random(size_h2) - 0.5
b3 = rg.random(size_out) - 0.5

b_arr = [b1, b2, b3]
```

```python
def ReLU(x):
    return np.maximum(0, x)

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)
```

```python
classifier(x_in, W_arr, b_arr, ReLU)
```
